# MindSync AI Exploration Notebook

This notebook is used for exploratory data analysis (EDA) and visualization of datasets related to smartphone usage patterns and digital wellbeing. The goal is to understand the data and prepare it for further analysis and model training.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

## Load Datasets
In this section, we will load the datasets from the processed directory.

```python
# Load processed datasets
data_path = '../datasets/processed/'
df = pd.read_csv(data_path + 'smartphone_usage.csv')

# Display the first few rows of the dataset
df.head()
```

## Data Overview
Let's check the structure and summary statistics of the dataset.

```python
# Check the shape of the dataset
df.shape
```

```python
# Display summary statistics
df.describe()
```

## Data Visualization
We will visualize the relationships between screen time, social media time, productivity score, and addiction risk.

```python
# Visualize screen time vs productivity score
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='screen_time', y='productivity_score', hue='addiction_risk', palette='coolwarm')
plt.title('Screen Time vs Productivity Score')
plt.xlabel('Screen Time (hours)')
plt.ylabel('Productivity Score')
plt.legend(title='Addiction Risk')
plt.show()
```

```python
# Visualize distribution of addiction risk
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='addiction_risk', palette='pastel')
plt.title('Distribution of Addiction Risk')
plt.xlabel('Addiction Risk')
plt.ylabel('Count')
plt.show()
```

## Conclusion
In this notebook, we performed exploratory data analysis on the smartphone usage dataset. We visualized key relationships and distributions that will inform our modeling efforts in predicting digital wellbeing scores and addiction risks.
